# **1**. What is the difference between a function and a method in Python?
 ## Function

A function is an independent block of code

- Defined outside a class

- Does not require an object

- Called directly by its name


```python
# example :
def multiply(a, b):
    return a * b

print(multiply(2, 4))
```

# Method

- A method is a function defined inside a class

- Called using an object

- Uses self as first parameter

- Works on object data


```python
#example
class Math:
    def multiply(self, a, b):
        return a * b

m = Math()
print(m.multiply(2, 4))



# note :
# Function → outside the class

# Method → inside the class

# Function → no object required

# Method → object required
```

# 6. Explain the concept of generators in Python and how they are defined.

## What is a Generator?

- A generator is a special type of function

- It generates values one at a time

- Uses less memory

- Uses yield instead of return

##  How are Generators Defined?

- Defined using def keyword

- yield keyword is used to return values

- Function call returns an iterator object

```python
# example :
def even_numbers(n):
    for i in range(2, n+1, 2):
        yield i

gen = even_numbers(6)

for num in gen:
    print(num)
```

# 7. What are the advantages of using generators over regular functions?
## Advantages of Generators

- Use less memory

- Produce values one at a time

- Ideal for large datasets

- Execution can pause and resume

- Faster and more efficient in long iteration

```python
# example:Regular Function
def square_list(n):
    return [i*i for i in range(n)]


print(square_list(4))
```

```python
# eample: Generator Function
def square_gen(n):
    for i in range(n):
        yield i*i


for s in square_gen(4):
    print(s)
```

# 9. Explain the purpose and usage of the `map()` function in Python.

- The map() function applies a function to each element

- Used to transform data in an iterable

- Makes code shorter and readable

```python
# Example: map() with function
def double(x):
    return x * 2


values = [1, 2, 3]
output = list(map(double, values))
print(output)
```

# 10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

## map() Function

- Applies a function to each element

- Used for data transformation

- Output size remains the same


```python
# Example (map()):

nums = [2, 4, 6]
result = list(map(lambda x: x + 1, nums))
print(result)
```

# filter() Function

- Filters elements based on a condition

- Returns only elements that satisfy condition

- Used for data selection

```python
#Example (filter()):

nums = [10, 15, 20, 25]
result = list(filter(lambda x: x > 15, nums))
print(result)
```

# reduce() Function

- Reduces all elements to one single value

- Requires functools.reduce

- Used for aggregation operations

```python
# Example (reduce()):

from functools import reduce


nums = [2, 3, 4]
result = reduce(lambda a, b: a * b, nums)
print(result)
```

```python
# 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in  the list

def sum_even(nums):
    total = 0
    for num in nums:
        if num % 2 == 0:  # check if number is even
            total += num
    return total

numbers = [10, 15, 20, 25,-12]
print("Sum of even numbers:", sum_even(numbers))

```

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

primes = [x for x in range(1, 201) if is_prime(x)]
print("Primes 1-200:", primes)
```

```python
# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        self.count += 1
        self.a, self.b = self.b, self.a + self.b
        return self.a

fib = Fibonacci(10) # inter number your choece
print(list(fib))
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

for p in powers_of_two(5):
    print(p, end=" ")
```

```python
# 7. Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_lines(file_path): # use erro handling becouse , filenotfounderror
    try:
        with open(file_path, 'r') as f:
            for line in f:
                yield line.strip()
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")

for line in read_file_lines('sample.txt'):
    print(line)
```
