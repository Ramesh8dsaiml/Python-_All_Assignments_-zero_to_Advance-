#1.  What is Object-Oriented Programming (OOP)

- Object-Oriented Programming (OOP) is a programming style where we organize code like real-world objects. It combines data (attributes) and functions (methods) into a single entity called an object.

    - Key Concepts:

    - Class: Blueprint for objects. Defines properties and behaviors.

    - Object: An instance of a class. Has actual data and methods.

     - Encapsulation: Keeps data and methods together, allows controlled access.

    - Inheritance: One class can inherit properties/methods from another class.

     - Polymorphism: Same function can behave differently for different objects.

     - Abstraction: Shows only necessary details, hides complexity.

     

```python
# example
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def start(self):
        print(f"{self.brand} {self.model} car is starting!")

my_car = Car("Toyota", "Corolla")
my_car.start()
```

# 4.What is the difference between abstraction and encapsulation

# Encapsulation

- Wrapping data and methods together and hiding the internal details.

- Use private variables and getter/setter methods.

- Focus: Data security and controlled access.

# Abstraction

- Showing only essential features and hiding implementation details.

- Use abstract classes or methods.

- Focus: Simplifying interface, hiding complexity.


```python
# example
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # private

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())  # 1500
```

```python
# example :abstraction
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car is starting")

c = Car()
c.start()
```

# 5. What are dunder methods in Python

- Dunder Methods = Double UNDERSCORE methods, also called magic methods.

- They customize Python’s built-in behavior.

- Syntax: __method_name__ (double underscores on both sides).

- Not called directly; Python calls them during special operations.

# Common Dunder Methods:

- __init__ → Constructor

- __str__ → For print

- __repr__ → Official representation

- __add__, __sub__ → Operators

- __len__ → Length

```python
# example :
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Point({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2  # __add__ is called
print(p3)     # __str__ is called
```

# 10. What are class and static methods in Python
### Class Method

- A method that works with the class, not the object.

- First parameter is cls (refers to the class).

- Defined using @classmethod.

- Use case:>>  Class-level operations, factory methods
# Static Method

- Does not take class or object reference.

- Works like a utility function.

- Defined using @staticmethod.



```python
# classmethod
class Student:
    school_name = "ABC School"

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name

print(Student.school_name)  # ABC School
Student.change_school("XYZ School")
print(Student.school_name)  # XYZ School
```

# 11. What is method overloading in Python
### Method Overloading means creating multiple versions of the same method with different number of parameters.

- Python does not support true method overloading because the latest method definition overwrites previous ones.

- But we can achieve similar behavior using default arguments or *args / **kwargs.
# note:

- Same method name add with different number of arguments → Simulates overloading

- Python overwrites previous definitions → *use default arguments or args

```python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(5))        # 5
print(calc.add(5, 10))    # 15
print(calc.add(5, 10, 15))# 30
```

# 12. What is method overriding in OOP
### Method Overriding means redefining a parent class method in the child class.

- The method in the child class changes the behavior of the parent method.

- This is an example of runtime polymorphism


```python
#
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):  # Method overriding
        print("Dog barks")

a = Animal()
a.speak()  # Animal makes a sound

d = Dog()
d.speak()  # Dog barks




# Animal → Parent class

# Dog → Child class

# Dog overrides speak() → Changes behavior
```

# 13 .What is a property decorator in Python
### Property Decorator (@property) is a Python feature that allows a method to be accessed like an attribute.

- Used for getter, setter, and deleter methods.

 - Makes encapsulation easier and more pythonic.
# Note:

- @property → Makes method like an attribute

- @age.setter → Allows controlled modification

- Benefit: Encapsulation + clean syntax

```python
class Student:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    @property
    def age(self):             # getter
        return self._age

    @age.setter
    def age(self, value):      # setter
        if value > 0:
            self._age = value
        else:
            print("Invalid age")

s = Student("Ramesh", 20)
print(s.age)
s.age = 25
print(s.age)
s.age = -5
```

#15.  What is an abstract class in Python
### An Abstract Class is a class that cannot be instantiated directly and usually contains one or more methods without implementation.

- In Python, we use ABC module and @abstractmethod decorator to create abstract classes.

- Child classes must implement all abstract methods.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):  # Abstract class
    @abstractmethod
    def start(self):
        pass  # Implementation in child class

class Car(Vehicle):
    def start(self):
        print("Car is starting")

c = Car()
c.start()  # Car is starting
```

# 18. What is multiple inheritance in Python

### Multiple Inheritance means a child class can inherit from more than one parent class.

- In Python, a child class can access properties and methods from multiple parent classes.

- If same method exists in multiple parents, Python uses Method Resolution Order (MRO) to decide which method to call.
 # note:
- Child → inherits from Father and Mother

- Child method overrides parent methods if defined

- Benefit: Combine multiple features in a single child class

```python
class Father:
    def skills(self):
        print("Gardening")

class Mother:
    def skills(self):
        print("Cooking")

class Child(Father, Mother):
    def skills(self):
        print("Programming")

c = Child()
c.skills()  # Programming
```

#19.  Explain the purpose of ‘’__str__’ and ‘__repr__’ ‘ methods in Python

### __repr__ Method:

- Provides official/unambiguous string representation of an object.

# 23. How does polymorphism work in Python with inheritance
### Polymorphism means “same interface, different behavior”.

- In Python, polymorphism works with inheritance via method overriding.

- When a child class overrides a parent class method, and a parent class reference points to the child object, the child’s method is executed → this is runtime polymorphism.

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Polymorphism in action
animals = [Dog(), Cat(), Animal()]

for a in animals:
    a.speak()
```

# 24. What is method chaining in Python OOP

## Method Chaining means calling multiple methods one after another on the same object in a single line.

- Achieved by returning the object (self) from each method.

- Use case: Fluent interface, readable code

```python
class Calculator:
    def __init__(self, value=0):
        self.value = value

    def add(self, x):
        self.value += x
        return self  # Return object

    def multiply(self, x):
        self.value *= x
        return self  # Return object

    def show(self):
        print(self.value)
        return self  # Return object

calc = Calculator()
calc.add(5).multiply(3).show()  # Method chaining
```

# 25.What is the purpose of the __call__ method in Python?
## The __call__ Method is a special method that allows an object to be called like a function.

- Meaning you can do object() and execute its behavior.
# note:

- __call__ → Makes object callable like a function

- Benefit: Flexible, readable, and functional-style code

```python
class Greeting:
    def __init__(self, name):
        self.name = name

    def __call__(self, message):
        print(f"{message}, {self.name}!")

greet = Greeting("Ramesh")
greet("Hello")  # Object called like a function
```

```python
# 2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle  from it and implement the area() method in both.

from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius**2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(4, 6)
print("Circle area:", c.area())
print("Rectangle area:", r.area())
```

```python
# 5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance.
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(200)
print("Balance:", acc.get_balance())
```

```python
# 14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary

class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        base_salary = super().calculate_salary()
        return base_salary + self.bonus

e1 = Employee("Ramesh", 40, 500)
m1 = Manager("Sita", 40, 500, 2000)

print(f"{e1.name}'s Salary: {e1.calculate_salary()}")
print(f"{m1.name}'s Salary: {m1.calculate_salary()}")
```

```python
# 16. Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that implement the sound() method

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Moo")

class Sheep(Animal):
    def sound(self):
        print("Baa")

c = Cow()
s = Sheep()

c.sound()  # Moo
s.sound()  # Baa
```
