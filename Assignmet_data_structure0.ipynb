# 14. How does the “in” keyword work differently for lists and dictionaries

### #The in keyword is used to check membership, but it works differently for lists and dictionaries.

1. in with List

- Checks whether a value exists as an element in the list

- Performs a linear search

2. in with Dictionary

- Checks whether a key exists in the dictionary

- Does not check values by default

- Uses hashing, so lookup is fast

# Note:
- In a list, the in keyword checks for values, while in a dictionary it checks only for keys.

```python
# example
nums = [10, 20, 30] # with list

print(20 in nums)
print(40 in nums)
```

```python
emp = {
    "id": 101,
    "name": "ramesh",
    "salary": 50000
}

print("name" in emp)  # Check for key
print("salary" in emp)
print("age" in emp)   # Key not present
print(101 in emp)     # Checks for key, not value
```

# 17. Describe the time complexity of accessing elements in a dictionary

#### #Accessing elements in a Python dictionary takes O(1) time on average due to hashing, with a rare worst-case time complexity of O(n).

1. Average Time Complexity

- O(1) — Constant time

- This is because dictionaries use a hash table

- The key is hashed and mapped directly to a memory location

2. Worst Case Time Complexity

- O(n) — Linear time

- Occurs when many keys collide into the same bucket

- This situation is rare due to Python’s optimized hashing

```python
# example
data = {"a": 10, "b": 20, "c": 30}
print(data["b"])   # O(1)
```

# 19. Why are dictionaries considered unordered, and how does that affect data retrieval

- Dictionaries are considered unordered because they store data using hashing rather than sequence, and this does not affect data retrieval since values are accessed directly using keys.

 # Why are dictionaries considered unordered?

- Dictionaries are considered unordered because:

- They are implemented using hash tables

- Keys are stored based on their hash values

- The internal order depends on hashing, not insertion position

- Even though modern Python versions preserve insertion order,
conceptually dictionaries are designed for fast lookup, not ordering.

# How does this affect data retrieval?

- Data retrieval is not affected

- Elements are accessed using keys, not positions

- Lookup time remains O(1) regardless of order


```python
# example
info = {"id": 101, "name": "ramesh"}
print(info["name"])
```
