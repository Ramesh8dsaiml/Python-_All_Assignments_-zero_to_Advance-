# 9. What is the difference between multithreading and multiprocessing
- Multithreading: Running multiple threads inside a single process.

- Threads share the same memory space

   -  Example: GUI updates + background tasks

- Multiprocessing: Running multiple processes, each with separate memory space.

- Processes are independent

   - Example: CPU-intensive tasks
   

````python
import threading
import time

def print_numbers():
    for i in range(5):
        print(i)
        time.sleep(1)

t1 = threading.Thread(target=print_numbers)
t1.start()
t1.join()
````

````python
from multiprocessing import Process

def print_numbers():
    for i in range(5):
        print(i)

p1 = Process(target=print_numbers)
p1.start()
p1.join()
````

# 15 .How does Python's garbage collection system work
- Garbage Collection Meaning:
In Python, garbage collection is the automatic process of freeing memory occupied by objects that are no longer in use.

- How it Works:

#  Reference Counting:

- Each object has a reference count.

- When the reference count drops to zero, Python frees the memory.

# Garbage Collector (GC) Module:

- Python’s gc module also handles cyclic references.

- Cyclic reference: When objects reference each other and reference count never becomes zero

# Automatic Cleanup:

- Python runs garbage collector in the background to free memory.

- Programmers usually don’t need to manually free memory.

````python
a = [1, 2, 3]
b = a      # Reference count increases
del a      # Reference count decreases
# Object still exists because b refers to it
del b      # Reference count = 0 → memory freed
````

````python
import gc

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

a = Node(1)
b = Node(2)
a.next = b
b.next = a  # cyclic reference

del a
del b

gc.collect()  # GC automatically frees cyclic objects
````

# 18. What is the difference between os.fork() and multiprocessing in Python

# os.fork()

- Works only on Unix/Linux systems.

- os.fork() creates a new child process which is an exact copy of the parent process.

- Child executes independently.

- Not supported on Windows.

- Low-level process control.

# Notes:

- Parent and child are separate processes.

- Need shared memory or pipes for communication.


# multiprocessing Module

- Cross-platform → works on Windows and Linux.

- High-level module, easier to create processes and manage them.

- Use Process class to create a new process.

- Provides easy IPC with Queue, Pipe, Manager.



````python
import os

pid = os.fork()
if pid == 0:
    print("Child process")
else:
    print("Parent process")
````

````python
from multiprocessing import Process

def func():
    print("Child process running")

p = Process(target=func)
p.start()
p.join()
print("Parent process finished")
````

# 19. What is the importance of closing a file in Python

### Why is closing a file important in Python?

1.  Release Memory:

- Open files use system memory.

- Closing the file frees that memory.

2.  Data Save (Flush):

- When writing to a file, data is stored in buffer first.

- Closing the file ensures buffer data is written to the file.

3.  Prevent File Corruption:

- If the file is not closed properly, data may become corrupted or incomplete.

4.  Resource Management:

- System resources like file descriptors are limited.

- Closing files frees resources for new files.

5.  Automatic Close with with Statement:

- Using with open(...) as f: ensures the file is automatically closed.

````python
# Normal way
f = open("example.txt", "w")
f.write("Hello Python")
f.close()  # Important!

# Using 'with' (automatic close)
with open("example.txt", "w") as f:
    f.write("Hello Python")
# No need to call f.close()
````

    - When objects reference each other.

   -  It may slow down program performance in some cases.

    - Programmers have limited direct control over memory allocation and deallocation.
